## 1 - Variables numéricas

Una variable se crea con `=`. Python infiere el tipo automáticamente. En análisis de datos los valores numéricos representan métricas: precios, conteos, tasas.

```python
precio_unitario = 15.99   # float: tiene parte decimal
cantidad        = 50       # int: entero, sin decimal

print(precio_unitario)
print(cantidad)
```

```python
ventas_enero   = 12500
ventas_febrero = 14800
ventas_marzo   = 11200

print(ventas_enero)
print(ventas_febrero)
print(ventas_marzo)
```

```python
temperatura_dia   = 36.5
temperatura_noche = 19.8

diferencia_temp = temperatura_dia - temperatura_noche
print(f'Diferencia de temperatura: {diferencia_temp}°C')
```

## 2 - Variables de texto (strings)

Los strings se crean con comillas simples o dobles. En datasets representan: nombres, categorías, IDs de texto, etiquetas.

```python
nombre_producto = 'Laptop Pro 15'
marca           = 'TechStore'

print(nombre_producto)
print(marca)
```

```python
nombre_ciudad = 'Bogotá'
pais          = 'Colombia'

# el operador + concatena strings
print('Ciudad: ' + nombre_ciudad + ', ' + pais)
```

```python
nombre_empleado = 'María Torres'
departamento    = 'Finanzas'
cargo           = 'Analista Senior'

print(nombre_empleado)
print(departamento)
print(cargo)
```

## 3 - Función `type()`

`type()` retorna el tipo de dato de una variable. Fundamental para depurar pipelines: una columna que parece numérica puede ser `str` si viene con símbolos del CSV.

```python
clientes_activos = 1500
tasa_conversion  = 0.032

print(type(clientes_activos))   # <class 'int'>
print(type(tasa_conversion))    # <class 'float'>
```

```python
precio  = 99.99
nombre  = 'Laptop'
activo  = True
stock   = 100

print('precio es de tipo:', type(precio))
print('nombre es de tipo:', type(nombre))
print('activo es de tipo:', type(activo))
print('stock  es de tipo:', type(stock))
```

```python
id_cliente = '98745'   # llega como string desde el CSV

print('Tipo original:', type(id_cliente))  # str

id_cliente_int = int(id_cliente)           # conversión a entero
print('Tipo convertido:', type(id_cliente_int))  # int
```

## 4 - Números: `int` y `float`

| Tipo | Descripción | Ejemplo |
|------|-------------|---------|
| `int` | Entero, sin decimales | `100`, `-5` |
| `float` | Con decimales | `3.14`, `-0.5` |

`int(3.9)` → `3` (trunca, no redondea). `float(5)` → `5.0`.

```python
ganancias        = 45780.50
ganancias_entero = int(ganancias)  # int() trunca la parte decimal

print('Con decimal:  ', ganancias)        # 45780.5
print('Sin decimal:  ', ganancias_entero)  # 45780  (se pierde .50)
```

```python
items_totales  = float(437)              # int → float
precio_promedio = items_totales / 3       # la division siempre devuelve float

print(precio_promedio)
print(type(precio_promedio))  # <class 'float'>
```

```python
stock_inicial  = 1000
stock_vendido  = 347
stock_restante = stock_inicial - stock_vendido

print(stock_restante)        # 653
print(type(stock_restante))  # int: int - int = int
```

## 5 - Operaciones matemáticas

| Operador | Descripción | Ejemplo |
|----------|-------------|---------|
| `+` `-` `*` | Suma, resta, multiplicación | `5 * 3 = 15` |
| `/` | División (siempre float) | `5 / 2 = 2.5` |
| `//` | División entera (trunca) | `5 // 2 = 2` |
| `%` | Módulo (el resto) | `5 % 2 = 1` |
| `**` | Potencia | `2 ** 8 = 256` |

```python
precio_unitario = 15.99
cantidad        = 50

precio_total        = precio_unitario * cantidad
precio_con_descuento = precio_total * (1 - 0.15)  # 15% de descuento

print(f'Precio total:          {precio_total:.2f}')
print(f'Precio con descuento:  {precio_con_descuento:.2f}')
```

```python
productos = 250
por_caja  = 12

cajas_completas = productos // por_caja   # division entera: descarta el resto
sobrantes       = productos %  por_caja   # modulo: solo el resto

print(f'Cajas completas: {cajas_completas}')
print(f'Productos sobrantes: {sobrantes}')
```

```python
ventas_este_ano  = 85000
ventas_ano_pasado = 72000

crecimiento = ((ventas_este_ano - ventas_ano_pasado) / ventas_ano_pasado) * 100
print(f'Crecimiento: {round(crecimiento, 2)}%')
```

## Resumen

| Concepto | Sintaxis |
|----------|----------|
| Crear variable | `precio = 29.99` |
| Ver tipo | `type(variable)` |
