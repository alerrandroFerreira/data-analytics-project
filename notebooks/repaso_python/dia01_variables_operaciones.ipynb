## 4. Números: `int` y `float`

| Tipo | Descripción | Ejemplo |
|------|-------------|---------|
| `int` | Entero, sin decimales | `100`, `-5` |
| `float` | Con decimales | `3.14`, `-0.5` |

Conversión entre tipos:
```python
int(3.9)    # → 3   (trunca, NO redondea)
float(5)    # → 5.0
```

## 5. Operaciones Matemáticas

| Operador | Descripción | Ejemplo |
|----------|-------------|---------|
| `+` `-` `*` | Suma, resta, multiplicación | `5 * 3 = 15` |
| `/` | División (siempre float) | `5 / 2 = 2.5` |
| `//` | División entera (trunca) | `5 // 2 = 2` |
| `%` | Módulo (el resto) | `5 % 2 = 1` |
| `**` | Potencia | `2 ** 8 = 256` |

## Resumen del Día 1

| Concepto | Sintaxis |
|----------|----------|
| Crear variable | `precio = 29.99` |
| Ver tipo | `type(variable)` |
| Convertir a entero | `int(3.7)` → `3` |

| Potencia | `2 ** 8` → `256` |
| Redondear | `round(x, 2)` |
