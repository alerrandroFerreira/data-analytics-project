## 1. Listas

Las listas son colecciones **ordenadas y mutables**. Se crean con `[]`.

```python
productos = ["Laptop", "Mouse", "Teclado"]

productos[0]       # "Laptop"       (primer elemento)
productos[-1]      # "Teclado"      (último elemento)
len(productos)     # 3
productos.append("Monitor")        # agrega al final
productos.pop(1)                   # elimina por índice
```

## 3. Diccionarios

Los diccionarios almacenan datos como pares **clave: valor**. Se crean con `{}`.
Son la estructura más cercana a una fila de una tabla.

```python
producto = {"nombre": "Monitor", "precio": 450.0, "stock": 15}
producto["nombre"]     # "Monitor"
producto["stock"] = 20 # modificar valor
producto.keys()        # todas las claves
producto.values()      # todos los valores
```

## 4. Booleanos

## Resumen del Día 3

| Estructura | Creación | Mutable | Acceso |
|------------|----------|---------|--------|
| Lista | `[1, 2, 3]` | ✓ | `lista[0]` |
| Tupla | `(1, 2, 3)` | ✗ | `tupla[0]` |
| Diccionario | `{"k": v}` | ✓ | `dic["k"]` |

| Control | Cuándo usarlo |
|---------|---------------|
| `if` | una sola condición |
| `if / else` | dos caminos |
| `if / elif / else` | múltiples caminos excluyentes |
