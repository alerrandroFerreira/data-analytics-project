Referencia de estructuras de datos y lógica condicional en Python. Estas estructuras son el equivalente directo de filas, columnas y filtros en pandas.

## 1 - Listas

Colecciones ordenadas y mutables. Se crean con `[]`. En análisis de datos modelan columnas, series de valores o registros temporales.

```python
productos = ['Laptop', 'Mouse', 'Teclado', 'Monitor', 'Webcam']

print(productos[0])       # primer elemento: 'Laptop'
print(productos[-1])      # último elemento:  'Webcam'
print(len(productos))     # total: 5
```

```python
precios = [15.99, 23.50, 8.75, 42.00, 19.99]

precios.append(55.00)   # agrega al final
precios.pop(2)          # elimina el elemento en índice 2 (8.75)

print(precios)
```

```python
ventas_diarias = [1200, 980, 1450, 2100, 890, 1670, 1340]

ventas_diarias.sort()          # ordena la lista en su lugar (no crea una nueva)
print('Ordenadas:', ventas_diarias)

maximo   = max(ventas_diarias)
minimo   = min(ventas_diarias)
promedio = sum(ventas_diarias) / len(ventas_diarias)

print(f'Máximo:   {maximo}')
print(f'Mínimo:   {minimo}')
print(f'Promedio: {promedio:.1f}')
```

## 2 - Tuplas

Colecciones ordenadas e inmutables. Se crean con `()`. Se usan cuando los datos no deben cambiar: coordenadas, configuración, registros fijos.

```python
coordenadas_tienda = (4.7110, -74.0721)

print('Latitud:', coordenadas_tienda[0])
print('Longitud:', coordenadas_tienda[1])
```

```python
registro_venta = ('Laptop', 1299.99, '2024-01-15', 'completada')

for i, valor in enumerate(registro_venta):
    print(f'[{i}] {valor}')

# intentar modificar una tupla produce TypeError
# registro_venta[1] = 999.99  → TypeError: 'tuple' object does not support item assignment
```

```python
datos_empleado = ('Carlos', 'Ventas', 45000)

# tuple unpacking: desempaca en variables individuales en una sola línea
nombre, departamento, salario = datos_empleado

print(nombre)
print(departamento)
print(salario)
```

## 3 - Diccionarios

Pares clave:valor. Se crean con `{}`. Son la estructura más cercana a una fila de tabla: cada clave es una columna.

```python
producto = {"nombre": 'Monitor 27"', "precio": 450.00, "stock": 15}

print(producto['nombre'])
print(producto['precio'])
print(producto['stock'])
```

```python
producto = {"nombre": 'Monitor 27"', "precio": 450.00, "stock": 15}

producto['categoria'] = 'Electrónica'   # agregar clave nueva
producto['precio']    = 420.00           # modificar valor existente

print(list(producto.keys()))
print(list(producto.values()))
```

```python
inventario = [
    {'nombre': 'Laptop',  'precio': 899.99, 'stock': 12},
    {'nombre': 'Monitor', 'precio': 450.00, 'stock': 8},
    {'nombre': 'Mouse',   'precio': 25.00,  'stock': 50},
]

# acceso: índice de lista → clave de diccionario
print(inventario[1]['nombre'])   # segundo producto: 'Monitor'
print(inventario[2]['precio'])   # precio del tercero: 25.0
```

## 4 - Booleanos

Valores `True` / `False`. Resultado de comparaciones. Base de todos los filtros de datos.

```python
en_stock         = True
descuento_activo = False

print(type(en_stock))
print(en_stock and descuento_activo)   # True AND False = False
print(en_stock or descuento_activo)    # True OR  False = True
print(not en_stock)                    # NOT True = False
```

```python
precio     = 150
presupuesto = 200

print(precio < presupuesto)   # True
print(precio == presupuesto)  # False
print(precio >= 100)          # True
print(precio != presupuesto)  # True
```

```python
# valores 'falsy': se comportan como False en un condicional
for val in [0, 1, '', 'hola', None, [], [1, 2]]:
    print(f'bool({repr(val):10}) = {bool(val)}')
```

## 5 - Estructura `if`

Ejecuta un bloque de código solo si la condición es `True`. La indentación (4 espacios) es obligatoria en Python.

```python
stock = 25

if stock > 0:
    print('Producto disponible')
else:
    print('Agotado')
```

```python
temperatura_sensor = 38.5

if temperatura_sensor >= 38:
    print(f'ALERTA: Temperatura de {temperatura_sensor}°C supera el umbral')
```

```python
ventas_del_dia = 15000
meta_diaria    = 12000

if ventas_del_dia > meta_diaria:
    print(f'Meta superada. Excedente: {ventas_del_dia - meta_diaria:,}')
else:
    print(f'Faltan {meta_diaria - ventas_del_dia:,} para cumplir la meta')
```

## 6 - `if / elif / else`

`elif` = 'else if'. Python evalúa las condiciones en orden y se detiene en la primera `True`. Usar `elif` (no `if` separados) garantiza que las ramas sean mutuamente excluyentes.

```python
puntaje_satisfaccion = 9

if puntaje_satisfaccion >= 9:
    print('Promotor')
elif puntaje_satisfaccion >= 7:
    print('Pasivo')
else:
    print('Detractor')
```

```python
precio_base    = 100
cantidad_pedido = 35

if cantidad_pedido < 10:
    descuento = 0
elif cantidad_pedido < 50:
    descuento = 0.05    # 5%
else:
    descuento = 0.10    # 10%

precio_final = precio_base * (1 - descuento) * cantidad_pedido
print(f'Precio final: {precio_final:.2f} (descuento {descuento*100:.0f}%)')
```

```python
for ingreso_anual in [15000, 35000, 70000, 120000]:
    if ingreso_anual < 20000:
        nivel = 'Nivel bajo'
    elif ingreso_anual < 50000:
        nivel = 'Nivel medio'
    elif ingreso_anual < 100000:
        nivel = 'Nivel alto'
    else:
        nivel = 'Nivel muy alto'
    print(f'{ingreso_anual:>8,} → {nivel}')
```

---
## Resumen

| Estructura | Creación | Mutable | Acceso |
|------------|----------|---------|--------|
| Lista | `[1, 2, 3]` | Sí | `lista[0]` |
| Tupla | `(1, 2, 3)` | No | `tupla[0]` |
| Diccionario | `{'k': v}` | Sí | `dic['k']` |

| Control | Cuándo usarlo |
|---------|---------------|
| `if` | una sola condición |
| `if / else` | dos caminos |
| `if / elif / else` | múltiples caminos excluyentes |
