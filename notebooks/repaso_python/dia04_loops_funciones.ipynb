## 1 - Loop `for`

Itera sobre cualquier colección: lista, string, rango, diccionario. En análisis de datos se usa para aplicar una transformación fila a fila antes de tener pandas.

```python
productos = ['Laptop', 'Mouse', 'Teclado', 'Monitor']

for producto in productos:
    print(producto)
```

```python
precios = [29.99, 15.50, 89.99, 199.99, 7.50]

total = 0
for precio in precios:
    total += precio    # acumulador: suma cada precio al total anterior

print(f'Total: {total:.2f}')
```

```python
ventas = [1200, 980, 2100, 1450, 890, 1670, 340]

contador = 0
for venta in ventas:
    if venta > 1000:
        contador += 1

pct = contador / len(ventas) * 100
print(f'Ventas > 1000: {contador} ({pct:.1f}%)')
```

## 2 - Función `range()`

`range(inicio, fin, paso)` genera una secuencia de enteros. El `fin` no se incluye. Se usa cuando se necesita iterar N veces o generar índices.

```python
for i in range(1, 8):       # 1, 2, 3, 4, 5, 6, 7
    print(f'Día {i}')
```

```python
for pct in range(0, 101, 10):   # 0, 10, 20, ..., 100
    print(f'Progreso: {pct}%')
```

```python
ids = []
for i in range(1, 6):                    # 1, 2, 3, 4, 5
    codigo = f'PROD_{str(i).zfill(3)}'  # zfill(3) rellena con ceros: '001', '002'...
    ids.append(codigo)

print(ids)
```

## 3 - Loop `while`

Repite mientras una condición sea `True`. Se usa cuando no se sabe de antemano cuántas iteraciones son necesarias. `break` fuerza la salida del loop.

```python
stock = 10

while stock > 0:
    stock -= 1    # equivale a stock = stock - 1
    print(f'Stock restante: {stock}')
```

```python
capital = 1000
tasa    = 0.08    # 8% anual
anio    = 0

while capital < 2000:
    capital = capital * (1 + tasa)
    anio   += 1
    print(f'Año {anio}: {capital:.2f}')

print(f'El capital se duplicó en {anio} años.')
```

```python
base_productos = ['Laptop', 'Mouse', 'Teclado', 'Monitor']

# while True + break: loop indefinido que solo termina desde adentro
# descomenta para ejecutar de forma interactiva:
# while True:
#     busqueda = input('Buscar (o escribe salir): ')
#     if busqueda == 'salir':
#         break
#     if busqueda in base_productos:
#         print('Encontrado')
#     else:
#         print('No encontrado')

# versión estática para demostración
for busqueda in ['Laptop', 'Impresora', 'Mouse', 'salir']:
    if busqueda == 'salir':
        break
    print('Encontrado' if busqueda in base_productos else 'No encontrado')
```

## 4 - Funciones

Las funciones encapsulan lógica reutilizable. En análisis de datos cada transformación repetida debería ser una función para poder aplicarla a columnas enteras.

```python
def calcular_iva(precio):
    return precio * 1.21    # IVA del 21%

print(calcular_iva(100))     # 121.0
print(calcular_iva(899.99))  # 1088.9879
print(calcular_iva(49.95))   # 60.4395
```

```python
def clasificar_venta(monto):
    if monto < 500:
        return 'Venta baja'
    elif monto < 2000:
        return 'Venta media'
    else:
        return 'Venta alta'

montos = [150, 500, 1800, 2000, 5000]
for m in montos:
    print(f'{m:>5}: {clasificar_venta(m)}')
```

```python
def resumen_ventas(lista_ventas):
    return {
        'total'   : sum(lista_ventas),
        'promedio': sum(lista_ventas) / len(lista_ventas),
        'maximo'  : max(lista_ventas),
        'minimo'  : min(lista_ventas),
        'cantidad': len(lista_ventas)
    }

ventas = [1500, 320, 2800, 950, 4200, 180, 760]
resultado = resumen_ventas(ventas)

for clave, valor in resultado.items():
    print(f'{clave:10}: {valor}')
```

## 5 - Funciones que llaman funciones

Las funciones pueden llamarse entre sí para construir pipelines: cada función hace una cosa y se combinan en funciones de nivel superior.

```python
def calcular_iva(precio):
    return precio * 1.21

def aplicar_descuento(precio, porcentaje):
    return precio * (1 - porcentaje / 100)

def precio_final(precio, porcentaje):
    precio_con_descuento = aplicar_descuento(precio, porcentaje)
    return calcular_iva(precio_con_descuento)    # llama a calcular_iva

print(f'Precio final (10% desc + IVA): {precio_final(100, 10):.2f}')
```

```python
def calcular_iva(precio):
    return precio * 1.21

def es_venta_valida(monto):
    return monto > 0

def procesar_venta(monto):
    if es_venta_valida(monto):          # llama a es_venta_valida
        return calcular_iva(monto)     # llama a calcular_iva
    return 'Venta inválida'

for monto in [150, 0, -50, 2000]:
    print(f'procesar_venta({monto:>5}) → {procesar_venta(monto)}')
```

```python
def limpiar_nombre(texto):
    return texto.strip().lower()

def formatear_codigo(texto):
    limpio = limpiar_nombre(texto)      # llama a limpiar_nombre
    return limpio.replace(' ', '_')

def generar_id(nombre, numero):
    codigo = formatear_codigo(nombre)   # llama a formatear_codigo
    return f'ID_{codigo}_{numero}'

print(generar_id('  Laptop Pro  ', 42))   # ID_laptop_pro_42
```

## Resumen

| Estructura | Cuándo usarla |
|------------|---------------|
| `for item in lista` | iterar sobre elementos conocidos |
| `for i in range(n)` | iterar N veces con índice |
| `while condicion` | iterar hasta que algo cambie |
| `while True` + `break` | loop hasta condición interna |

| Función | Patrón |
|---------|--------|
| Básica | `def f(x): return x` |
| Con condición | `def f(x): if ... return ... else return ...` |
| Retorna dict | `def f(x): return {'k': v}` |
| Llama a otra | `def f(x): return g(x) + 1` |
