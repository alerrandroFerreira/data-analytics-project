## 2. Función `range()`

`range()` genera una secuencia de números.

```python
range(5)          # 0, 1, 2, 3, 4
range(1, 6)       # 1, 2, 3, 4, 5
range(0, 101, 10) # 0, 10, 20, ..., 100
```

`range(inicio, fin, paso)` — el `fin` no se incluye.

## 3. Loop `while`

El `while` repite mientras una condición sea `True`.
Usa `break` para salir manualmente.

```python
contador = 0
while contador < 5:
    print(contador)
    contador += 1    # ← siempre avanzar, evitar loop infinito
```

Se usa cuando no sabes de antemano cuántas iteraciones necesitas.

## 5. Funciones que Llaman Funciones

Las funciones pueden llamar a otras funciones.
Esto permite construir **pipelines** donde cada función hace una cosa y las combinas.

```python
def paso_1(x):
    return x * 2

def paso_2(x):
    resultado = paso_1(x)   # llama a paso_1
    return resultado + 10

paso_2(5)   # → 20
```

## Resumen del Día 4

| Estructura | Cuándo usarla |
|------------|---------------|
| `for item in lista` | iterar sobre elementos conocidos |
| `for i in range(n)` | iterar N veces con índice |
| `while condicion` | iterar hasta que algo cambie |
| `while True` + `break` | loop hasta condición interna |

| Función | Patrón |
|---------|--------|
| Básica | `def f(x): return x` |
| Con condición | `def f(x): if ... return ... else return ...` |
| Retorna dict | `def f(x): return {"k": v}` |
| Llama a otra | `def f(x): return g(x) + 1` |
