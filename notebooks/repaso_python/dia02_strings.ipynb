## 1 - Métodos básicos de strings

Los strings son inmutables. Los métodos devuelven un nuevo string, no modifican el original.

```python
texto.upper()   # TODO EN MAYUSCULAS
texto.lower()   # todo en minusculas
texto.strip()   # elimina espacios al inicio y final
len(texto)      # longitud
```

````python
descripcion = 'Análisis de Ventas Q1 2024'

print(len(descripcion))         # longitud total del string
print(descripcion.upper())      # para normalizar comparaciones
print(descripcion.lower())      # para nombres de columna
````

````python
email = '  maria.gonzalez@empresa.com  '   # espacios deliberados

print('Antes:', len(email))
email_limpio = email.strip()               # elimina espacios de bordes
print('Después:', len(email_limpio))
print(email_limpio)
````

````python
nombre_columna = '  Total Ventas USD  '

# encadenar métodos: primero strip, luego lower, luego replace
# el resultado de cada método se pasa como entrada al siguiente
nombre_normalizado = nombre_columna.strip().lower().replace(' ', '_')
print(nombre_normalizado)   # 'total_ventas_usd'
````

## 2 - Indexar strings

Cada carácter tiene un índice que empieza en `0`. Los negativos cuentan desde el final.

```
texto = 'Python'
         P y t h o n
índice:  0 1 2 3 4 5
negat.: -6-5-4-3-2-1
```

````python
producto = 'Smartphone Galaxy S24'

print(producto[0])    # primer carácter: 'S'
print(producto[-1])   # último carácter:  '4'
print(producto[5])    # posición 5: 'h'
````

````python
codigo = 'PRD-2024-001'

print(codigo[-3:])    # últimos 3 caracteres: '001'
print(codigo[-4])     # posición -4: '-'
````

````python
ruta = 'datos/ventas/enero_2024.csv'

# forma manual: contar posiciones hasta la 'e' de 'enero'
print(ruta[12])                # 'e' de 'enero'

# forma automática: .index() devuelve la primera posición del substring
print(ruta.index('enero'))     # 12
````

## 3 - Slicing

El slicing extrae una porción del string: `texto[inicio:fin:paso]`.
El índice `fin` no se incluye en el resultado.

````python
fecha = '2024-01-15'

anio = fecha[0:4]    # '2024'
mes  = fecha[5:7]    # '01'
dia  = fecha[8:10]   # '15'

print(f'Año:  {anio}')
print(f'Mes:  {mes}')
print(f'Día:  {dia}')
````

````python
codigo = 'PROD_CAT_001_MX'

parte1 = codigo[0:4]    # 'PROD'
parte2 = codigo[5:8]    # 'CAT'
parte3 = codigo[9:12]   # '001'
parte4 = codigo[13:]    # 'MX'

print(parte1, parte2, parte3, parte4)
````

````python
texto = 'ventas_totales_2024'

invertido = texto[::-1]    # tercer parámetro -1: recorre al revés
print('Invertido:', invertido)

anio  = texto[-4:]         # últimos 4 caracteres
print('Año:', anio)

ventas = texto[:6]         # primeros 6 caracteres
print('Palabra:', ventas)
````

## 5 - Formateo de strings (f-strings)

Los f-strings insertan variables en el texto. Formatos numéricos útiles:
- `:.2f` → 2 decimales
- `:,` → separador de miles
- `:,.2f` → ambos

````python
nombre_producto = 'Monitor 4K'
precio          = 899.99

print(f'Producto: {nombre_producto} | Precio: ${precio}')
````

````python
ventas_totales = 1528763.50

# :,.2f combina separador de miles y 2 decimales
print(f'Ventas del mes: ${ventas_totales:,.2f}')
````

````python
empresa      = 'RetailPro'
total_ventas = 3250000
fecha        = 'Q1 2024'

# \n dentro del f-string produce salto de línea
print(f'=== REPORTE {empresa} ===\nPeríodo: {fecha}\nTotal ventas: ${total_ventas:,.2f}')
````
