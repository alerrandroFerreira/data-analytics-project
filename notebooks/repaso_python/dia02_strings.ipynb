## 3. Filtrar Strings (Slicing)

El slicing extrae una porción del string: `texto[inicio:fin:paso]`

```python
texto = "2024-01-15"
texto[0:4]    # "2024"   (del 0 al 3, el 4 no se incluye)
texto[5:7]    # "01"
texto[:4]     # "2024"   (desde el inicio)
texto[5:]     # "01-15"  (hasta el final)
texto[::-1]   # string al revés
```

## 5. Formateo de Strings (f-strings)

Los f-strings insertan variables directamente en el texto.

```python
nombre = "Ana"
ventas = 1500000.75

print(f"Analista: {nombre}")
print(f"Ventas: ${ventas:,.2f}")   # → Ventas: $1,500,000.75
```

Formatos numéricos útiles dentro de `{}`:
- `:.2f` → 2 decimales
- `:,` → separador de miles
- `:,.2f` → ambos
