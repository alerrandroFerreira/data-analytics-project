```python
from pathlib import Path
import pandas as pd

def find_project_root():
    current = Path.cwd()
    for parent in [current] + list(current.parents):
        if (parent / '.git').exists() or (parent / 'data').exists():
            return parent
    return current

PROJECT_ROOT = find_project_root()
DATA_EXTERNAL = PROJECT_ROOT / 'data' / 'external'

print(f"Check: Proyecto raíz: {PROJECT_ROOT}")
print(f"Check: Datos external: {DATA_EXTERNAL}")
```

```python
import pandas as pd
df = pd.read_csv(DATA_EXTERNAL / 'CovidNumbers.csv')
df.head()  
```

```python
import matplotlib.pyplot as plt
#pyplot crea y gestiona varios temas sobre la vision y manera en que se muestran los datos
```

```python
#Crea histograma, que contiene del dataframe la columna Monthly Rate, para saber por ejemplo 
# si la mayoria de tasas son altas o bajas
plt.hist(df["Monthly Rate"])
plt.title("Monthly rate of Covid")
plt.xlabel("Monthly Rate")
plt.ylabel("Frecuencia")
plt.show()
```

```python
#convertir el formato de datos del date a formato real de fecha
df["Year Month"] = pd.to_datetime(df["Year Month"].astype(int).astype(str), format="%Y%m")
```

```python
#dibuja puntos usando open en X y close en Y.
plt.scatter(df["Year Month"], df["Monthly Rate"])
plt.title("Tasa mensual a lo largo del tiempo")
plt.xlabel("Año-Mes")
plt.ylabel("Monthly Rate")
plt.show()
```
