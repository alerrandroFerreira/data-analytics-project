```python
from pathlib import Path
import pandas as pd

def find_project_root():
    current = Path.cwd()
    for parent in [current] + list(current.parents):
        if (parent / '.git').exists() or (parent / 'data').exists():
            return parent
    return current

PROJECT_ROOT = find_project_root()
DATA_EXTERNAL = PROJECT_ROOT / 'data' / 'external'

print(f"Check: Proyecto raíz: {PROJECT_ROOT}")
print(f"Check: Datos external: {DATA_EXTERNAL}")
```

```python
df = pd.read_csv(DATA_EXTERNAL /'USDC.csv')
df
```

```python
#convertir el formato de datos del date a formato real de fecha
df["date"] = pd.to_datetime(df["date"])
```

```python
import matplotlib.pyplot as plt
```

```python
#crea un histograma usando la columna close, dividiendo los datos en 20 intervalos, 
# para ver en que frecuencia aparecen
plt.hist(df["close"], bins=20)
#se añade un titulo
plt.title("Distribución del precio de cierre")
#etiqueta horizontal que mustra rangos de precios
plt.xlabel("Precio")
#representa la vertical, cuantas veces aparece cada rango de precio
plt.ylabel("Frecuencia")
#muestra el gráfico en pantalla
plt.show()
```

```python
#se seleccionan las columnas relevantes para hacer la correlación
correlation = df[["open", "high", "low", "close"]].corr()
print(correlation)
#muestra la matriz como imagen
plt.imshow(correlation)
#añade una escala de colores a la representacion
plt.colorbar()
plt.xticks(range(4), ["open","high","low","close"])
plt.yticks(range(4), ["open","high","low","close"])
plt.title("Correlación")
plt.show()
```

```python
#dibuja puntos usando open en X y close en Y.
plt.scatter(df["open"], df["close"])
plt.title("Open vs Close")
plt.xlabel("Open")
plt.ylabel("Close")
plt.show()
```
