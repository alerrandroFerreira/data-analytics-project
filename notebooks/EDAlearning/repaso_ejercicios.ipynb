## Encargo 2 — Limpieza profesional Airbnb

> *"El equipo de Airbnb nos pasó su export de listings. Necesito que lo limpies siguiendo un criterio profesional, no solo quitando nulos a lo bruto. Dame un resumen de qué decidiste en cada paso y por qué."*

**2.1** Carga `Listings.csv` con los parámetros correctos para que no falle la codificación. Imprime shape y tipos de dato.

**2.2** Aplica el orden correcto de limpieza. Para cada paso, justifica la decisión:
- Corrige los tipos de `host_since` y las columnas booleanas (`host_is_superhost`, etc.)
- Elimina o conserva duplicados
- Trata los nulos de `district` (86% nulos), `bedrooms` (10% nulos) y `host_response_rate` (46% nulos)
- Decide qué hacer con los nulos en `review_scores_rating` y justifica por qué no los rellenas

**2.3** Normaliza `host_response_time`: elimina espacios y convierte a minúsculas. Imprime los valores únicos antes y después.

**2.4** Crea la columna `amenities_count` a partir de la columna `amenities`. Imprime la media y el máximo de esa nueva columna.

```python
import pandas as pd

from pathlib import Path

def find_project_root():
    for parent in [Path.cwd()]+list(Path.cwd().parents):
        if (parent / 'data').exists():
            return parent
ROOT = find_project_root()
df = pd.read_csv(ROOT / 'data' / 'external' / 'Listings.csv', encoding='latin-1', low_memory=False)

df = df.copy()


print(f'Su shape: {df.shape}')
# Corrección de tipos
df['host_since'] = pd.to_datetime(df['host_since'])#,format='%d/%m/%Y'
print(f'Corrección de formatos: {df['host_since'].dtypes}')

col_bool = ['instant_bookable','host_is_superhost','host_has_profile_pic','host_identity_verified']
for col in col_bool:
    df[col] = df[col].map({'t': True, 'f': False})
print(df[col_bool].dtypes)

n_dup = df.duplicated().sum()
print(f'Filas duplicadas: {n_dup}')

if n_dup>0:
    df = df.drop_duplicates()
    print(f'Shape tras eliminar Duplicados: {df.shape}')
else:
    print('No existen duplicados')

#------------------------------------- Buen metodo para identificar porcentaje de nulos
nulos = (df.isnull().mean()*100).round(3)
print('Porcentaje de nulos por columna:')
print(nulos[nulos>0].to_string())
#-------------------------------------
UMBRAL_NULOS=0.50
col_drop_null = [
    c for c in df.columns 
    if df[c].isnull().mean()>UMBRAL_NULOS
    ]
print(f'Columnas con >{UMBRAL_NULOS*100:.0f}% nulos: {col_drop_null}')
df = df.drop(columns=col_drop_null)
print(f'Shape tras eliminar columnas vacías: {df.shape}')


```
