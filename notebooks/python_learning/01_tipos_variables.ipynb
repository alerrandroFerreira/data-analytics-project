# 01 — Tipos, variables y el modelo de objetos

## Tipado dinámico

Python infiere el tipo en tiempo de ejecución. No existe `int x = 5`, `String nombre = ''` ni nada similar.

```python
precio   = 15.99        # float — Python lo sabe por el punto decimal
cantidad = 50            # int
activo   = True          # bool
nombre   = 'Monitor 4K' # str

# type() devuelve la clase del objeto — en Python todo es objeto
for var in [precio, cantidad, activo, nombre]:
    print(f'{var!r:<20} → {type(var).__name__}')
```

## Todo es un objeto

`id()` devuelve la dirección de memoria del objeto. `is` compara identidad, `==` compara valor.

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a           # c apunta al mismo objeto que a

print(a == b)   # True  — mismo valor
print(a is b)   # False — objetos distintos en memoria
print(a is c)   # True  — mismo objeto

c.append(4)
print(a)        # [1, 2, 3, 4] — a también cambió porque c y a son el mismo objeto
```

## Mutabilidad

Objetos inmutables: `int`, `float`, `str`, `tuple`, `bool`, `frozenset`.
Objetos mutables: `list`, `dict`, `set`.

La distinción importa al pasar argumentos a funciones y al usar objetos como claves de diccionario.

```python
# Inmutable — cada operación crea un objeto nuevo
x = 'ventas'
print(id(x))
x = x.upper()   # no modifica el original, crea 'VENTAS'
print(id(x))    # id diferente — es un objeto nuevo

# Mutable — las operaciones modifican el objeto existente
nums = [10, 20, 30]
print(id(nums))
nums.append(40)
print(id(nums))  # mismo id — el objeto se modificó en sitio
```

## None

`None` es el equivalente de `null` / `nil` / `undefined`. Hay un único objeto `None` en memoria.

```python
resultado = None

# Comparar None siempre con 'is', no con '=='
if resultado is None:
    print('sin resultado todavía')

# None es falsy — funciona en condiciones directas también
if not resultado:
    print('falsy')

# Las funciones que no hacen return explícito devuelven None
def sin_return():
    x = 1 + 1

print(sin_return())  # None
```

## Conversiones de tipo

Python no convierte automáticamente entre tipos (a diferencia de JS). La conversión es explícita.

```python
# Desde CSV todo llega como string — la conversión es obligatoria
precio_str   = '15.99'
cantidad_str = '50'

total = float(precio_str) * int(cantidad_str)
print(f'Total: {total}')

# int() trunca, no redondea
print(int(9.99))    # 9
print(round(9.99))  # 10

# bool() — valores falsy: 0, 0.0, '', [], {}, None, set()
for v in [0, 0.0, '', [], {}, None, 'texto', 1, [0]]:
    print(f'{str(v):<12} → {bool(v)}')
```

## Type hints

Python 3.5+ soporta anotaciones de tipo. Son documentación — no se validan en tiempo de ejecución por defecto. `mypy` sí las valida.

```python
def calcular_margen(precio_venta: float, precio_costo: float) -> float:
    return (precio_venta - precio_costo) / precio_venta

# Python 3.10+ — union types con |
def parsear_id(valor: int | str) -> str:
    return str(valor).zfill(6)  # rellena con ceros a la izquierda hasta 6 chars

print(calcular_margen(100, 60))   # 0.4
print(parsear_id(42))             # '000042'
print(parsear_id('1234'))         # '001234'
```

## Múltiple asignación y swap

Python permite asignar varias variables en una línea y hacer swap sin variable temporal.

```python
# Asignación múltiple — las expresiones del lado derecho se evalúan primero
x, y = 10, 20
x, y = y, x     # swap sin temp
print(x, y)     # 20 10

# Útil para desempaquetar resultados de funciones
minimo, maximo = min([3,1,4,1,5]), max([3,1,4,1,5])
print(minimo, maximo)

# Asignación encadenada — todos apuntan al mismo objeto
a = b = c = 0
print(a is b is c)  # True — cuidado con objetos mutables
```
