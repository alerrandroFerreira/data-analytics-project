## enumerate

Devuelve `(índice, elemento)` en cada iteración. Elimina el antipatrón `range(len(lista))`.

```python
productos = ['Laptop', 'Mouse', 'Monitor', 'Teclado']

# Antipatrón — acceder al índice con range(len(...))
# for i in range(len(productos)):
#     print(i, productos[i])

# Forma Pythónica
for i, producto in enumerate(productos):
    print(f'{i+1}. {producto}')

# start= para empezar desde un número diferente
for n, p in enumerate(productos, start=100):
    print(f'SKU-{n}: {p}')
```

## zip

Itera en paralelo sobre múltiples iterables. Se detiene en el más corto.

```python
nombres  = ['Laptop', 'Mouse', 'Monitor']
precios  = [899.99, 15.50, 350.00]
stocks   = [12, 200, 45]

for nombre, precio, stock in zip(nombres, precios, stocks):
    print(f'{nombre:<12} ${precio:>8.2f}  stock: {stock}')

# Crear diccionario con zip
inventario = dict(zip(nombres, stocks))
print(inventario)

# zip_longest — rellena con fillvalue cuando los iterables tienen distinto tamaño
from itertools import zip_longest
a = [1, 2, 3, 4]
b = ['x', 'y']
print(list(zip_longest(a, b, fillvalue=None)))
```

## Cláusula else en loops

El bloque `else` se ejecuta si el loop terminó de forma natural (sin `break`). Patrón poco conocido pero muy útil para búsquedas.

```python
# Buscar un producto con stock crítico
inventario = {'Laptop': 12, 'Mouse': 200, 'Monitor': 45}

umbral_critico = 5
for producto, stock in inventario.items():
    if stock <= umbral_critico:
        print(f'Stock crítico: {producto}')
        break
else:
    print('Todos los productos tienen stock aceptable')

# El else no se ejecuta si hubo break — permite saber si la búsqueda fue exitosa
```

## itertools — herramientas para iteración

```python
from itertools import chain, islice, product, groupby

# chain — concatenar iterables sin crear listas intermedias
enero  = [120, 98, 145]
febrero = [200, 175, 189]
print(list(chain(enero, febrero)))

# islice — cortar un iterable sin materializarlo completo (lazy)
def generar_ids():
    n = 1
    while True:
        yield f'ID-{n:04d}'
        n += 1

print(list(islice(generar_ids(), 5)))  # primeros 5 IDs

# product — producto cartesiano (equivale a loops anidados)
regiones    = ['North', 'South']
categorias  = ['A', 'B', 'C']
combinaciones = list(product(regiones, categorias))
print(combinaciones)
```

## Generators vs listas

Un generator no calcula todos los valores de golpe — los produce uno a uno cuando se le pide. Útil con datasets grandes.

```python
# Lista: todos los valores en memoria de una vez
lista_cuadrados = [x**2 for x in range(1_000_000)]

# Generator expression: misma sintaxis con () — calcula bajo demanda
gen_cuadrados = (x**2 for x in range(1_000_000))

# next() extrae el siguiente valor
print(next(gen_cuadrados))   # 0
print(next(gen_cuadrados))   # 1

# sum/max/min/any/all aceptan generators directamente
ventas = [120, 98, 210, 145, 89]
total_grandes = sum(v for v in ventas if v > 100)   # sin crear lista intermedia
print(total_grandes)
```

---
## Resumen

| Patrón | Sintaxis |
|--------|----------|
| Índice + valor | `for i, v in enumerate(lista)` |
| Paralelo | `for a, b in zip(lista1, lista2)` |
| Buscar con flag | `for x in iter: ... else: no encontrado` |
| Concatenar | `chain(iter1, iter2)` |
| Lazy | `(expr for x in iter)` |
