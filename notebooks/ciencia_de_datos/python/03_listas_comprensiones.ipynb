# 03 — Listas y comprensiones

Las listas son secuencias mutables y ordenadas. A diferencia de arrays en otros lenguajes, pueden mezclar tipos y su tamaño es dinámico.

```python
# Creación
precios  = [15.99, 23.50, 8.75, 42.00]
mixta    = [1, 'Monitor', True, None, [2, 3]]  # válido pero poco habitual

# Métodos mutables — modifican la lista en sitio y devuelven None
precios.append(55.00)        # agrega al final
precios.extend([7.99, 3.50]) # agrega múltiples elementos
precios.insert(0, 99.99)     # inserta en posición específica

print(precios)
print(f'Longitud: {len(precios)}')

# pop() devuelve el elemento eliminado
ultimo   = precios.pop()     # elimina y devuelve el último
primero  = precios.pop(0)    # elimina y devuelve el índice 0
print(f'Eliminados: {ultimo}, {primero}')
```

## Slicing avanzado

Same syntax as strings — `lista[inicio:fin:paso]`. Slicing crea una copia superficial.

```python
ventas = [120, 98, 210, 145, 89, 167, 134, 201, 178, 95]

print(ventas[2:5])      # [210, 145, 89]  — índices 2, 3, 4
print(ventas[-3:])      # últimos 3
print(ventas[::2])      # elementos en índices pares
print(ventas[::-1])     # invertida

# Reemplazar un rango con otro rango (de diferente tamaño)
ventas[1:3] = [999, 888, 777]   # reemplaza 2 elementos con 3
print(ventas)

# Copia superficial con slice
copia = ventas[:]          # equivalente a ventas.copy()
print(copia is ventas)     # False — objetos distintos
```

## sort vs sorted

`sort()` modifica la lista original y devuelve `None`. `sorted()` devuelve una lista nueva.

```python
datos = [
    {'nombre': 'Laptop',  'precio': 899, 'stock': 12},
    {'nombre': 'Mouse',   'precio': 15,  'stock': 200},
    {'nombre': 'Monitor', 'precio': 350, 'stock': 45},
]

# sorted() con key — sin modificar 'datos'
por_precio = sorted(datos, key=lambda x: x['precio'], reverse=True)
for item in por_precio:
    print(f"{item['nombre']:<12} ${item['precio']:>6}")

# Ordenar por múltiples criterios: primero por categoría, luego por precio
productos = [('B', 30), ('A', 50), ('A', 20), ('B', 10)]
print(sorted(productos, key=lambda x: (x[0], x[1])))
```

## List comprehensions

Reemplazan el patrón `for` → `append` con una expresión en una línea. La expresión va primero, el `for` después.

```python
precios = [15.99, 23.50, 8.75, 42.00, 5.50, 99.99]

# Versión con loop tradicional
con_iva = []
for p in precios:
    con_iva.append(round(p * 1.21, 2))

# Versión con comprensión — equivalente exacta
con_iva = [round(p * 1.21, 2) for p in precios]
print(con_iva)

# Con filtro — el if va al final
caros = [p for p in precios if p > 20]
print(caros)

# Transformación + filtro en una sola línea
caros_con_iva = [round(p * 1.21, 2) for p in precios if p > 20]
print(caros_con_iva)
```

## Comprensiones anidadas

Equivalen a loops `for` anidados. El orden de los `for` es el mismo que en los loops.

```python
# Matriz 3×4
matriz = [[i * j for j in range(1, 5)] for i in range(1, 4)]
for fila in matriz:
    print(fila)

# Aplanar una lista de listas (muy común en análisis de datos)
ventas_regiones = [[120, 98, 145], [200, 175, 189], [90, 110, 78]]
todas_ventas = [v for region in ventas_regiones for v in region]
print(todas_ventas)
print(f'Total: {sum(todas_ventas)}, Media: {sum(todas_ventas)/len(todas_ventas):.1f}')
```

## Unpacking extendido

```python
# El * captura el 'resto' — puede ir en cualquier posición
primero, *resto = [1, 2, 3, 4, 5]
print(primero)  # 1
print(resto)    # [2, 3, 4, 5]

*inicio, ultimo = [1, 2, 3, 4, 5]
print(inicio)   # [1, 2, 3, 4]
print(ultimo)   # 5

primero, *medio, ultimo = [1, 2, 3, 4, 5]
print(medio)    # [2, 3, 4]

# Útil para ignorar columnas que no interesan
nombre, _, precio, *_ = ['Monitor', 'SKU-001', '350.00', 'stock:15', 'cat:electronics']
print(nombre, precio)  # Monitor  350.00
```

## Resumen

| Patrón | Sintaxis |
|--------|----------|
| Crear | `[expr for x in iterable]` |
| Filtrar | `[expr for x in iterable if cond]` |
| Aplanar | `[v for sub in lista for v in sub]` |
| Unpacking | `a, *resto = lista` |
| Copia | `lista[:]` o `lista.copy()` |
