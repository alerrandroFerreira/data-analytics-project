# 02 — Strings

Los strings en Python son objetos inmutables. Cada método devuelve un string nuevo — el original no cambia.

```python
texto = '  Monitor 4K Ultra HD  '

# Métodos de limpieza — encadenables porque cada uno devuelve un nuevo str
limpio = texto.strip().title()
print(repr(texto))   # el original intacto
print(repr(limpio))

# Otros métodos de normalización
codigo = 'PROD-2024-MX'
print(codigo.lower())           # 'prod-2024-mx'
print(codigo.replace('-', '_')) # 'PROD_2024_MX'
print(codigo.startswith('PROD')) # True
print(codigo.endswith('MX'))     # True
print(codigo.count('-'))         # 2
```

## Indexado y slicing

`str[inicio:fin:paso]` — el índice `fin` no se incluye. Índices negativos cuentan desde el final.

```python
fecha = '2024-11-28'

anio = fecha[:4]     # '2024'
mes  = fecha[5:7]    # '11'
dia  = fecha[8:]     # '28'

# Invertir un string — paso -1 recorre al revés
print(fecha[::-1])   # '82-11-4202'

# Slicing nunca lanza IndexError — ajusta al rango disponible
s = 'hola'
print(s[0:100])      # 'hola' — sin error
# s[10]              # IndexError — acceso por índice sí puede fallar
```

## split y join

`split` convierte string → lista. `join` convierte lista → string. Son operaciones inversas.

```python
linea_csv = 'Monitor 4K,899.99,15,Electronics'

campos = linea_csv.split(',')        # ['Monitor 4K', '899.99', '15', 'Electronics']
nombre, precio, stock, categoria = campos   # unpacking directo

# join: el separador va antes de .join(), la lista va dentro
campos_limpios = [c.strip() for c in campos]
reconstruido   = ' | '.join(campos_limpios)
print(reconstruido)

# splitlines() para texto con saltos de línea
multilinea = 'ene\nfeb\nmar'
print(multilinea.splitlines())  # ['ene', 'feb', 'mar']
```

## f-strings

La forma más directa de formatear en Python 3.6+. Soportan expresiones completas, no solo variables.

```python
empresa    = 'RetailPro'
ventas     = 3_250_000.5    # _ como separador visual — solo Python
crecimiento = 0.1274

# :,.2f → separador de miles + 2 decimales
# :.1% → porcentaje con 1 decimal
# !r    → representación con repr() — útil para debugging
print(f'{empresa}: ${ventas:,.2f} ({crecimiento:.1%} vs año anterior)')

# Expresiones dentro de {}
precios = [10, 25, 30, 45]
print(f'Media: {sum(precios)/len(precios):.2f}')

# Alineación: > derecha, < izquierda, ^ centro
for producto, precio in [('Laptop', 999), ('Mouse', 15), ('Monitor', 350)]:
    print(f'{producto:<12} {precio:>8,.0f} €')
```

## find, in y partition

```python
ruta = 'datos/ventas/enero_2024.csv'

# 'in' es la forma Pythónica de comprobar existencia — más legible que find() >= 0
print('ventas' in ruta)       # True

# find() devuelve el índice o -1 — útil cuando necesitas la posición
pos = ruta.find('/')
print(pos)                    # 5
print(ruta[pos+1:])           # 'ventas/enero_2024.csv'

# partition() divide en 3 partes: antes, separador, después
antes, sep, despues = ruta.partition('/')
print(antes)    # 'datos'
print(despues)  # 'ventas/enero_2024.csv'

# rpartition() busca desde el final — ideal para extensiones
nombre, _, extension = 'reporte_2024.csv'.rpartition('.')
print(nombre, extension)   # reporte_2024  csv
```

## Strings multilínea

```python
# Triple comilla — conserva saltos de línea y sangría
query = """
    SELECT region, SUM(sales) AS total
    FROM ventas
    WHERE year = 2024
    GROUP BY region
"""

# textwrap.dedent elimina la sangría común si molesta
import textwrap
print(textwrap.dedent(query).strip())
```

## Resumen

| Operación | Sintaxis |
|-----------|----------|
| Limpiar espacios | `s.strip()` |
| Dividir | `s.split(',')` |
| Unir | `', '.join(lista)` |
| Buscar | `'x' in s` o `s.find('x')` |
| Formatear | `f'{var:,.2f}'` |
| Slicing | `s[inicio:fin:paso]` |
