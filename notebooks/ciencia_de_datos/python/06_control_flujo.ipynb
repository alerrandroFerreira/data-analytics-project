## if / elif / else y operadores booleanos

`and`, `or`, `not` son palabras clave en Python (no `&&`, `||`, `!`). El cortocircuito aplica igual.

```python
stock = 25
precio = 89.99
cliente_premium = True

# Cortocircuito: si la primera condición de 'and' es False, la segunda no se evalúa
if stock > 0 and precio < 100:
    print('Disponible y asequible')

# Encadenar comparaciones — Pythónico, equivale a (10 < stock < 100)
if 10 < stock < 100:
    print(f'Stock moderado: {stock}')

# or para valores por defecto (short-circuit)
nombre = '' or 'Anónimo'   # '' es falsy — devuelve 'Anónimo'
precio_final = None or 0.0
print(nombre, precio_final)
```

## Expresión ternaria

Equivalente al operador `? :` de C/Java pero con sintaxis más legible.

```python
stock = 0

# valor_si_true if condicion else valor_si_false
estado = 'disponible' if stock > 0 else 'agotado'
print(estado)

# Dentro de f-strings
for precio in [5, 50, 500]:
    segmento = 'bajo' if precio < 20 else 'medio' if precio < 200 else 'alto'
    print(f'{precio:>4} → {segmento}')
```

## any() y all()

Evitan escribir `or`/`and` encadenados sobre iterables.

```python
stocks = {'Laptop': 0, 'Mouse': 200, 'Monitor': 5}

# any — True si al menos uno cumple la condición
hay_agotado = any(s == 0 for s in stocks.values())
print(f'Algún agotado: {hay_agotado}')

# all — True si todos cumplen la condición
todos_disponibles = all(s > 0 for s in stocks.values())
print(f'Todos disponibles: {todos_disponibles}')

# any/all con iterable vacío
print(any([]))   # False — ninguno cumple (vacío)
print(all([]))   # True  — todos cumplen (vacío, no hay contraejemplo)
```

## Walrus operator `:=`

Asigna y evalúa en la misma expresión (Python 3.8+). Útil en condiciones con efectos secundarios.

```python
import re

# Sin walrus — calcular y luego comprobar
lineas = ['ventas: 1200', 'nota: revisar', 'ventas: 890', 'costo: 450']

for linea in lineas:
    match = re.search(r'ventas: (\d+)', linea)
    if match:
        print(f'Venta encontrada: {match.group(1)}')

# Con walrus — asigna y comprueba en una línea
for linea in lineas:
    if m := re.search(r'ventas: (\d+)', linea):
        print(f'Venta encontrada: {m.group(1)}')
```

## match / case (Python 3.10+)

Más expresivo que `if/elif` para comparar un valor contra múltiples patrones. Soporta destructuring.

```python
def clasificar_pedido(estado: str, monto: float) -> str:
    match estado:
        case 'pendiente' if monto > 1000:
            return 'pedido grande — revisar'
        case 'pendiente':
            return 'en cola'
        case 'enviado' | 'entregado':
            return 'completado'
        case 'cancelado':
            return 'cancelado'
        case _:
            return f'estado desconocido: {estado}'

for estado, monto in [('pendiente', 1500), ('pendiente', 50), ('enviado', 100), ('X', 0)]:
    print(f'{estado:<12} ${monto:<6} → {clasificar_pedido(estado, monto)}')
```

## Resumen

| Patrón | Sintaxis |
|--------|----------|
| Ternario | `x if cond else y` |
| Al menos uno | `any(expr for x in iter)` |
| Todos | `all(expr for x in iter)` |
| Asignar+evaluar | `if val := expresion:` |
| Multi-patrón | `match x: case 'a' \| 'b':` |
