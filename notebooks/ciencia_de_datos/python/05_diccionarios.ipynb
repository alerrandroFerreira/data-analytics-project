# 05 — Diccionarios

La estructura de datos central en Python. Equivale a HashMap/Map/Object en otros lenguajes. Desde Python 3.7, el orden de inserción está garantizado.

```python
# Creación
producto = {
    'nombre':    'Monitor 4K',
    'precio':    450.00,
    'stock':     15,
    'activo':    True,
    'tags':      ['4K', 'IPS', '27 pulgadas'],
}

# Acceso — [] lanza KeyError si no existe, .get() devuelve None o default
print(producto['nombre'])              # 'Monitor 4K'
print(producto.get('descuento'))       # None
print(producto.get('descuento', 0.0))  # 0.0 — default explícito

# Modificación
producto['precio'] = 399.00
producto['color']  = 'Negro'   # agrega clave nueva
del producto['activo']          # elimina una clave
print(producto)
```

## Iterar sobre diccionarios

```python
inventario = {'Laptop': 12, 'Mouse': 200, 'Monitor': 45, 'Teclado': 78}

# .items() devuelve pares (clave, valor) — la forma más común
for producto, stock in inventario.items():
    estado = 'BAJO' if stock < 50 else 'OK'
    print(f'{producto:<12} {stock:>4}  [{estado}]')

print(list(inventario.keys()))    # ['Laptop', 'Mouse', 'Monitor', 'Teclado']
print(list(inventario.values()))  # [12, 200, 45, 78]
```

## setdefault y update

```python
# setdefault — inserta solo si la clave no existe, y devuelve el valor actual
ventas_por_region = {}

datos = [('West', 1200), ('East', 800), ('West', 1500), ('East', 950)]
for region, monto in datos:
    ventas_por_region.setdefault(region, []).append(monto)

print(ventas_por_region)

# update — merge de diccionarios (claves existentes se sobrescriben)
base  = {'nombre': 'Mouse', 'precio': 15}
extra = {'precio': 12, 'color': 'Negro', 'stock': 200}
base.update(extra)
print(base)

# Python 3.9+ — operador | para merge sin modificar el original
resultado = base | {'descuento': 0.10}
print(resultado)
```

## Dict comprehensions

```python
precios = {'Laptop': 899, 'Mouse': 15, 'Monitor': 350, 'Teclado': 65}

# Aplicar IVA a todos los precios
con_iva = {prod: round(p * 1.21, 2) for prod, p in precios.items()}
print(con_iva)

# Filtrar — solo productos por encima de 100
caros = {prod: p for prod, p in precios.items() if p > 100}
print(caros)

# Invertir clave-valor
invertido = {v: k for k, v in precios.items()}
print(invertido)

# Crear diccionario desde dos listas con zip
nombres = ['Laptop', 'Mouse', 'Monitor']
stocks  = [12, 200, 45]
inventario = dict(zip(nombres, stocks))
print(inventario)
```

## collections.Counter y defaultdict

```python
from collections import Counter, defaultdict

# Counter — cuenta ocurrencias, acepta cualquier iterable
ventas = ['MX', 'CO', 'MX', 'AR', 'MX', 'CO', 'BR', 'AR', 'MX']
conteo = Counter(ventas)
print(conteo)                      # Counter({'MX': 4, 'CO': 2, 'AR': 2, 'BR': 1})
print(conteo.most_common(3))       # top 3

# Combinar dos contadores
ventas_feb = Counter(['MX', 'BR', 'BR', 'CO'])
print(conteo + ventas_feb)

# defaultdict — evita KeyError al acceder a clave inexistente
por_categoria = defaultdict(list)
transacciones = [('Electronics', 900), ('Furniture', 400), ('Electronics', 150)]
for cat, monto in transacciones:
    por_categoria[cat].append(monto)
print(dict(por_categoria))
```

## Resumen

| Operación | Sintaxis |
|-----------|----------|
| Acceso seguro | `d.get(k, default)` |
| Insertar si no existe | `d.setdefault(k, default)` |
| Merge | `d1 \| d2` (Python 3.9+) |
| Comprensión | `{k: v for k, v in d.items()}` |
| Contar | `Counter(iterable)` |
| Acumular | `defaultdict(list)` |
