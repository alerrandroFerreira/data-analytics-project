# 04 — Tuplas, Sets y Frozensets

## Tuplas

Colecciones ordenadas e inmutables. Se usan cuando los datos no deben modificarse: coordenadas, registros, claves compuestas, retorno múltiple de funciones.

```python
# Creación — los paréntesis son opcionales, la coma define la tupla
punto        = (4.7110, -74.0721)
registro     = 'Laptop', 1299.99, '2024-01-15'   # válido sin paréntesis
singleton    = (42,)    # la coma es obligatoria para tuplas de un elemento

# Inmutable — esto lanza TypeError
# punto[0] = 5.0

# Pero si contiene un objeto mutable, ese objeto sí puede cambiar
t = ([1, 2], [3, 4])
t[0].append(99)    # no cambia la tupla, cambia la lista dentro
print(t)           # ([1, 2, 99], [3, 4])
```

## Unpacking y retorno múltiple

```python
# Desempaquetar coordenadas
lat, lon = (4.7110, -74.0721)

# Las funciones que devuelven múltiples valores devuelven una tupla
def estadisticas(datos: list) -> tuple[float, float, float]:
    return min(datos), max(datos), sum(datos) / len(datos)

ventas = [1200, 980, 1450, 2100, 890, 1670]
minimo, maximo, media = estadisticas(ventas)
print(f'min={minimo}, max={maximo}, media={media:.1f}')

# Tuplas como claves de diccionario — los tipos inmutables son hasheables
region_mes = {}
region_mes[('West', 'Ene')] = 45000
region_mes[('East', 'Ene')] = 38000
print(region_mes)
```

## Sets

Colecciones sin orden y sin duplicados. La pertenencia (`in`) es O(1) — mucho más rápida que en listas.

```python
# Set literal con {} — pero {} vacío crea dict, no set
categorias = {'Electronics', 'Furniture', 'Electronics', 'Office'}
print(categorias)   # {'Electronics', 'Furniture', 'Office'} — duplicado eliminado

# set() para crear desde lista (forma más común para deduplicar)
paises = ['MX', 'CO', 'AR', 'MX', 'CO', 'BR']
unicos = set(paises)
print(sorted(unicos))   # ['AR', 'BR', 'CO', 'MX']

# in es O(1) en set, O(n) en lista
paises_prioritarios = {'MX', 'CO', 'AR', 'BR', 'CL', 'PE'}
print('MX' in paises_prioritarios)    # True
print('US' in paises_prioritarios)    # False
```

## Operaciones de conjuntos

```python
clientes_enero  = {'A001', 'A002', 'A003', 'A004', 'A005'}
clientes_febrero = {'A003', 'A004', 'A005', 'A006', 'A007'}

# Union — todos los clientes que compraron en alguno de los dos meses
print(clientes_enero | clientes_febrero)

# Intersección — compraron en ambos meses (retención)
print(clientes_enero & clientes_febrero)

# Diferencia — compraron en enero pero no en febrero (perdidos)
print(clientes_enero - clientes_febrero)

# Diferencia simétrica — compraron solo en uno de los dos
print(clientes_enero ^ clientes_febrero)

# issubset — ¿A es subconjunto de B?
vip = {'A003', 'A004'}
print(vip.issubset(clientes_enero))    # True
```

## Frozenset

Set inmutable — sirve como clave de diccionario o como elemento de otro set.

```python
# Frozenset: mismo comportamiento que set pero inmutable
combo_A = frozenset({'Electronics', 'Furniture'})
combo_B = frozenset({'Furniture', 'Electronics'})   # mismo contenido

print(combo_A == combo_B)   # True — los sets no tienen orden

# Útil como clave de diccionario (los sets normales no son hasheables)
descuentos_por_combo = {
    frozenset({'Electronics', 'Office'}): 0.10,
    frozenset({'Furniture', 'Electronics'}): 0.15,
}

carrito = frozenset({'Office', 'Electronics'})
print(descuentos_por_combo.get(carrito, 0))   # 0.10
```

## Resumen

| Estructura | Orden | Mutable | Duplicados | Hasheable |
|------------|-------|---------|------------|-----------|
| list | sí | sí | sí | no |
| tuple | sí | no | sí | sí (si todo el contenido lo es) |
| set | no | sí | no | no |
| frozenset | no | no | no | sí |
