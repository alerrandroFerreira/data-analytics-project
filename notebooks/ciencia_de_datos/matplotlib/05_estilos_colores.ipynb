# 05 — Estilos, colores y personalización

El aspecto visual importa cuando se comunican resultados. Matplotlib tiene un sistema completo de temas, paletas y control de cada elemento gráfico.

## Setup

```python
%matplotlib widget
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import pandas as pd
import numpy as np
from pathlib import Path

def find_project_root():
    for parent in [Path.cwd()] + list(Path.cwd().parents):
        if (parent / 'data').exists():
            return parent

ROOT  = find_project_root()
TRAIN = ROOT / 'data' / 'external' / 'train.csv'
AIR   = ROOT / 'data' / 'external' / 'Listings.csv'

df = pd.read_csv(TRAIN, low_memory=False)
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
```

## Estilos predefinidos — `plt.style.use()`

```python
# Ver todos los estilos disponibles
print(plt.style.available)
```

```python
estilos = ['default', 'seaborn-v0_8', 'ggplot', 'fivethirtyeight']
datos   = df.groupby(df['Order Date'].dt.to_period('M').astype(str))['Sales'].sum()

fig, axes = plt.subplots(2, 2, figsize=(12, 7))

for ax, estilo in zip(axes.flat, estilos):
    with plt.style.context(estilo):   # aplicar estilo solo a este bloque
        ax.plot(datos.index, datos.values, linewidth=2)
        ax.set_title(f'Estilo: {estilo}', fontsize=10)
        ax.tick_params(axis='x', rotation=45)

fig.suptitle('Comparación de estilos', fontsize=13)
plt.tight_layout()
plt.show()
```

## Colores en matplotlib

Matplotlib acepta colores en múltiples formatos: nombre, hex, RGB, RGBA, o ciclos de la paleta.

```python
fig, ax = plt.subplots(figsize=(9, 4))

# Distintas formas de especificar color
x = np.arange(5)
ax.bar(x - 0.3, [1, 3, 2, 4, 3], width=0.25, color='steelblue',         label='nombre')
ax.bar(x - 0.0, [2, 2, 3, 2, 4], width=0.25, color='#E07B39',            label='hex')
ax.bar(x + 0.3, [3, 1, 4, 1, 2], width=0.25, color=(0.2, 0.6, 0.4, 0.8), label='RGBA')

ax.set_xticks(x)
ax.set_xticklabels(['A', 'B', 'C', 'D', 'E'])
ax.legend()
ax.set_title('Formas de especificar color')
ax.grid(axis='y', linestyle=':', alpha=0.4)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

## Colormaps — escala de color para valores numéricos

```python
ventas_region_mes = (
    df.groupby(['Region', df['Order Date'].dt.year])['Sales']
    .sum()
    .unstack('Region')
)

fig, ax = plt.subplots(figsize=(7, 3))

im = ax.imshow(
    ventas_region_mes.values,
    cmap='Blues',          # paleta: Blues, Reds, viridis, RdYlGn, ...
    aspect='auto',
)

ax.set_xticks(range(len(ventas_region_mes.columns)))
ax.set_xticklabels(ventas_region_mes.columns)
ax.set_yticks(range(len(ventas_region_mes.index)))
ax.set_yticklabels(ventas_region_mes.index)
ax.set_title('Ventas por región y año (intensidad = revenue)')

fig.colorbar(im, ax=ax, label='USD')

plt.tight_layout()
plt.show()
```

## `rcParams` — configuración global

```python
# rcParams controla todos los defaults de matplotlib en la sesión
# Útil para definir un estilo consistente al inicio del notebook

plt.rcParams.update({
    'figure.dpi':        120,
    'axes.spines.top':   False,   # ocultar borde superior
    'axes.spines.right': False,   # ocultar borde derecho
    'axes.grid':         True,
    'grid.linestyle':    ':',
    'grid.alpha':        0.4,
    'font.size':         10,
})

# Ahora todos los gráficos del notebook heredan estos defaults
fig, ax = plt.subplots(figsize=(8, 4))
ventas = df.groupby(df['Order Date'].dt.to_period('M').astype(str))['Sales'].sum()
ax.plot(ventas.index, ventas.values, color='steelblue', linewidth=2)
ax.set_title('Con rcParams personalizados')
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
# Restaurar defaults
plt.rcParams.update(plt.rcParamsDefault)
```

## Resumen

| Concepto | Sintaxis |
|----------|----------|
| Aplicar estilo global | `plt.style.use('seaborn-v0_8')` |
| Estilo temporal | `with plt.style.context('ggplot'): ...` |
| Color por nombre | `color='steelblue'` |
| Color hex | `color='#E07B39'` |
| Color RGBA | `color=(r, g, b, alpha)` |
| Colormap continua | `cmap='Blues'` en imshow/scatter |
| Configuración global | `plt.rcParams.update({...})` |
| Ver estilos disponibles | `plt.style.available` |
